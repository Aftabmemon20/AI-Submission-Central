```python
# Quick install and launch
!pip install -q gradio

import gradio as gr
import requests

def call_agent(text, agent_type):
    prompts = {
        "summarize": f"Summarize this medical text in 3-4 sentences:\n\n{text}",
        "sanitize": f"Remove ALL PHI (names, dates, IDs) from:\n\n{text}",
        "article": f"Write a research article outline about: {text}"
    }

    response = requests.post(
        "http://localhost:11434/api/generate",
        json={"model": "llama3.2:3b", "prompt": prompts[agent_type], "stream": False}
    )
    return response.json()['response']

# Create interface
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🏥 Medical AI Multi-Agent System")

    with gr.Tab("📝 Summarize"):
        sum_in = gr.Textbox(label="Medical Text", lines=8)
        sum_btn = gr.Button("Summarize", variant="primary")
        sum_out = gr.Textbox(label="Summary", lines=8)
        sum_btn.click(lambda x: call_agent(x, "summarize"), sum_in, sum_out)

    with gr.Tab("🔒 Remove PHI"):
        phi_in = gr.Textbox(label="Text with PHI", lines=8)
        phi_btn = gr.Button("Sanitize", variant="primary")
        phi_out = gr.Textbox(label="Sanitized", lines=8)
        phi_btn.click(lambda x: call_agent(x, "sanitize"), phi_in, phi_out)

    with gr.Tab("✍️ Write Article"):
        art_in = gr.Textbox(label="Topic", lines=3)
        art_btn = gr.Button("Generate", variant="primary")
        art_out = gr.Textbox(label="Article", lines=12)
        art_btn.click(lambda x: call_agent(x, "article"), art_in, art_out)

demo.launch(share=True)
```
